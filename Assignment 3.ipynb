```python
#Program to find whether or not a given triangle is a right angled triangle or not.
def right_triangle(a, b, c):
   if (a**2 == b**2 + c**2 or b**2 == a**2 + c**2 or c**2 == a**2 + b**2):
    print ("Right Angled Triangle")
   else:
    print ("Not a Right Angled Triangle")

x = float(input("Enter the first side length: "))
y = float(input("Enter the second side length: "))
z = float(input("Enter the third side length: "))

right_triangle(x, y, z)
right_triangle(7, 8, 9)
right_triangle(5, 4, 3)
right_triangle(3, 6, 9)
```
