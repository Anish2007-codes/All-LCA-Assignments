```python
import re

def check_string(s):
    pattern = "^[A-Za-z0-9]+$" # ^ = start, $ = end, + = one or more
    return bool(re.match(pattern, s))

# Example 
s1 = "Hello123"
s2 = "Hello_123" # contains underscore

print (check_string(s1)) # True
print (check_string(s2)) # False
```

```python
# Alternatively this code can also be written in short

def check_string(s):
    return s.isalnum()

# Example
s1 = "Hello123"
s2 = "Hello 123!"

print (check_string(s1)) # True
print (check_string(s2)) # False
```
